```python
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Ship Date"] = pd.to_datetime(df["Ship Date"])
print(df[["Order Date", "Ship Date"]].dtypes)
```

```python
df.info()
```

```python
df.groupby("Category")["Sales"].sum().sort_values(ascending=False).plot(kind="bar")
```

```python
import matplotlib.pyplot as plt
df.groupby("Category")["Sales"].sum().sort_values(ascending=False).plot(kind="bar")
plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.show()
```

```python
category_profit = df.groupby("Category")["Profit"].sum()
category_profit.plot(kind="bar", title="Total Profit by Category",xlabel="Category",ylabel="Total Profit")
```

```python
plt.show()
```

```python
subcategory_sales = df.groupby("Sub-Category")["Sales"].sum().sort_values(ascending=False)
subcategory_sales.plot(kind="bar",figsize=(10, 5),title="Total Sales by Sub-Category",xlabel="Sub-Category",ylabel="Total Sales" )
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(df["Sales"], df["Profit"], alpha=0.5)
plt.title("sales vs Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.show()
```

## Key Insights

1. Technology generated the highest total sales among the three main categories.
2. Technology also generated the highest total profit, while Furniture had the lowest profit.
3. Phones and Chairs were the top-selling sub=categories based on total sales.
4. Fasteners had the lowest sales among the sub-categories.
5. The Sales vs Profit analysis shows that higher sales generally tend to be associated with higher profit, but some high-sales orders still resulted in losses.
6. The dataset contains no missing values and no duplicate records, making it suitable for further analysis.

## Conclusion
