```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("hospital data analysis.csv")
print("Dataset loaded successfully!")
print("Dataset Shape:", df.shape)
df.head()
```

```python
print("Dataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nColumn Names:")
print(df.columns.tolist())
```

```python
df.describe()
```

```python
print("Patient Conditions:")
print(df["Condition"].value_counts())

print("\nGender Distribution:")
print(df["Gender"].value_counts())

print("\nReadmission Status:")
print(df["Readmission"].value_counts())
```

```python
plt.figure(figsize=(12, 6))

df["Condition"].value_counts().plot(
    kind="bar",
    color="skyblue"
)

plt.title("Distribution of Patient Conditions")
plt.xlabel("Medical Condition")
plt.ylabel("Number of Patients")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(data=df, x="Readmission")

plt.title("Hospital Patient Readmission Analysis")
plt.xlabel("Readmitted")
plt.ylabel("Number of Records")

plt.show()
```

```python
plt.figure(figsize=(12, 6))

avg_cost = df.groupby("Condition")["Cost"].mean().sort_values(ascending=False)

avg_cost.plot(kind="bar", color="orange")

plt.title("Average Treatment Cost by Medical Condition")
plt.xlabel("Medical Condition")
plt.ylabel("Average Cost")

plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize = (8, 5))
sns.countplot(data=df, x="Outcome")
plt.title("Patient Outcome Analysis")
plt.xlabel("Patient Outcome")
plt.ylabel("Number of Records")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x="Satisfaction")

plt.title("Patient Satisfaction Analysis")
plt.xlabel("Satisfaction Rating")
plt.ylabel("Number of Records")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))

avg_stay = df.groupby("Condition")["Length_of_Stay"].mean()
avg_stay = avg_stay.sort_values(ascending=False)

avg_stay.plot(kind="bar", color="mediumseagreen")

plt.title("Average Length of Stay by Medical Condition")
plt.xlabel("Medical Condition")
plt.ylabel("Average Length of Stay")

plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(data=df, x="Gender")

plt.title("Gender-wise Patient Analysis")
plt.xlabel("Gender")
plt.ylabel("Number of Records")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.histplot(data=df, x="Age", bins=10, kde=True)

plt.title("Patient Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Records")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))

avg_satisfaction = df.groupby("Condition")["Satisfaction"].mean()
avg_satisfaction = avg_satisfaction.sort_values(ascending=False)

avg_satisfaction.plot(kind="bar", color="purple")

plt.title("Average Patient Satisfaction by Medical Condition")
plt.xlabel("Medical Condition")
plt.ylabel("Average Satisfaction Rating")

plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(data=df, x="Length_of_Stay", y="Cost")

plt.title("Treatment Cost vs Length of Stay")
plt.xlabel("Length of Stay")
plt.ylabel("Treatment Cost")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 6))

numeric_df = df.select_dtypes(include="number")
correlation = numeric_df.drop(columns=["Patient_ID"]).corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap of Hospital Data")
plt.tight_layout()
plt.show()
```
